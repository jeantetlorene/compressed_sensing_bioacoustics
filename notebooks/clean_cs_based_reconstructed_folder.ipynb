```python
from pathlib import Path
```

```python
# --- Config ---
source_dir = Path(r"C:\Users\loren\Documents\Postdoc\Compressed_sensing\Data\Gibbon\Compressed_Audio\cs_0.4")
recon_dir = Path(r"C:\Users\loren\Documents\Postdoc\Compressed_sensing\Data\Gibbon\Compressed_Audio\cs_reconstructed_0.4")

SOURCE_SUFFIX = "_compressed.wav"
RECON_SUFFIX = "_reconstructed.npy"

dry_run = False  # set to False to actually delete
```

```python
# --- Build set of already-reconstructed base IDs ---
reconstructed_ids = {
    f.name[: -len(RECON_SUFFIX)]
    for f in recon_dir.iterdir()
    if f.name.endswith(RECON_SUFFIX)
}
print(f"Found {len(reconstructed_ids)} already-reconstructed file(s) in {recon_dir}")

# --- Find matching files to delete in source_dir ---
to_delete = [
    f for f in source_dir.iterdir()
    if f.name.endswith(SOURCE_SUFFIX) and f.name[: -len(SOURCE_SUFFIX)] in reconstructed_ids
]
print(f"Found {len(to_delete)} file(s) in {source_dir} to delete\n")

# --- Delete (or preview) ---
for f in to_delete:
    if dry_run:
        print(f"[DRY RUN] Would delete: {f.name}")
    else:
        f.unlink()
        print(f"Deleted: {f.name}")

if dry_run and to_delete:
    print("\nThis was a dry run. Set dry_run = False and re-run to actually delete.")
```
